## Step 1 — Load dan Validasi Dataset

Pada tahap ini, dataset hasil preprocessing dimuat dan diperiksa untuk memastikan struktur data sudah sesuai sebelum dilakukan feature engineering.

Pemeriksaan meliputi jumlah dokumen, nama kolom, nilai kosong, serta ketersediaan kolom `teks_clean` sebagai sumber utama feature engineering.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Load dataset hasil preprocessing
df = pd.read_csv('hasil_preprocessed.csv')

# Informasi dasar dataset
print("=== INFORMASI DATASET ===")
print(f"Jumlah dokumen : {len(df)}")
print(f"Jumlah kolom   : {len(df.columns)}")
print(f"Kolom          : {list(df.columns)}")

# Cek nilai kosong pada kolom penting
print("\n=== CEK NILAI KOSONG ===")
print(df[['no', 'judul', 'topik', 'teks_clean']].isnull().sum())

# Tampilkan beberapa data
print("\n=== CONTOH DATA ===")
display(df[['no', 'judul', 'topik', 'teks_clean']].head())
```

## Step 2 — Menentukan Teks sebagai Input Feature Engineering

Feature engineering dilakukan menggunakan kolom `teks_clean`, yaitu teks yang telah melalui tahap preprocessing pada Exercise 4.

Kolom `no`, `judul`, dan `topik` tidak digunakan sebagai fitur TF-IDF karena berfungsi sebagai metadata dokumen.

```python
# Mengambil teks hasil preprocessing sebagai corpus
corpus = df['teks_clean'].fillna('').astype(str)

print("Jumlah dokumen dalam corpus:", len(corpus))
print("Contoh teks bersih:")
print(corpus.iloc[0][:500])
```

## Step 3 — Ekstraksi Fitur dengan TF-IDF

TF-IDF digunakan untuk mengubah teks menjadi representasi numerik berdasarkan tingkat kepentingan suatu term dalam dokumen dan seluruh corpus.

Pada tahap awal digunakan kombinasi unigram dan bigram agar fitur tidak hanya berupa kata tunggal, tetapi juga dapat menangkap kombinasi dua kata yang memiliki makna lebih spesifik.

```python
# Inisialisasi TF-IDF Vectorizer
tfidf_vectorizer = TfidfVectorizer(
    lowercase=False,
    min_df=2,
    max_df=0.85,
    ngram_range=(1, 2)
)

# Transformasi teks menjadi matriks TF-IDF
X_tfidf = tfidf_vectorizer.fit_transform(corpus)

# Mendapatkan nama fitur
feature_names = tfidf_vectorizer.get_feature_names_out()

print("=== HASIL TF-IDF BASELINE ===")
print(f"Jumlah dokumen : {X_tfidf.shape[0]}")
print(f"Jumlah fitur   : {X_tfidf.shape[1]}")
print(f"Ukuran matriks : {X_tfidf.shape}")
```

## Step 4 — Inspeksi Fitur dan Bobot TF-IDF

Pada tahap ini dilakukan pemeriksaan terhadap fitur yang terbentuk dan bobot TF-IDF pada salah satu dokumen.

Analisis ini digunakan untuk melihat term atau kombinasi term yang paling dominan dalam merepresentasikan sebuah artikel.

```python
# Ambil vektor TF-IDF artikel pertama
doc_0_vector = X_tfidf[0].toarray()[0]

# Gabungkan nama fitur dengan bobotnya
df_top_features = pd.DataFrame({
    'fitur': feature_names,
    'bobot_tfidf': doc_0_vector
})

# Urutkan berdasarkan bobot terbesar
df_top_features = df_top_features.sort_values(
    by='bobot_tfidf',
    ascending=False
)

print("Judul artikel:")
print(df['judul'].iloc[0])

print("\nTopik:")
print(df['topik'].iloc[0])

print("\n10 fitur dengan bobot TF-IDF tertinggi:")
display(df_top_features.head(10).reset_index(drop=True))
```

## Step 5 — Refinement Fitur TF-IDF

Refinement dilakukan untuk mengurangi fitur yang terlalu jarang atau terlalu umum serta mengontrol dimensi matriks TF-IDF.

Berbeda dengan preprocessing, refinement pada tahap ini tidak mengubah teks sumber, tetapi menyaring fitur yang digunakan dalam representasi numerik.

```python
# TF-IDF refined
tfidf_refined = TfidfVectorizer(
    lowercase=False,
    min_df=3,
    max_df=0.70,
    ngram_range=(1, 2),
    max_features=1500
)

# Transformasi corpus
X_tfidf_refined = tfidf_refined.fit_transform(corpus)

# Nama fitur refined
feature_names_refined = tfidf_refined.get_feature_names_out()

print("=== HASIL REFINEMENT TF-IDF ===")
print(f"Jumlah dokumen : {X_tfidf_refined.shape[0]}")
print(f"Jumlah fitur   : {X_tfidf_refined.shape[1]}")
print(f"Ukuran matriks : {X_tfidf_refined.shape}")
```

## Step 6 — Perbandingan TF-IDF Baseline dan Refined

Hasil baseline dan refined dibandingkan berdasarkan jumlah fitur yang terbentuk.

Perbandingan ini digunakan untuk melihat pengaruh penyaringan fitur terhadap dimensi representasi dokumen.

```python
# Jumlah fitur
jumlah_fitur_baseline = X_tfidf.shape[1]
jumlah_fitur_refined = X_tfidf_refined.shape[1]

# Menghitung persentase pengurangan fitur
pengurangan = (
    (jumlah_fitur_baseline - jumlah_fitur_refined)
    / jumlah_fitur_baseline
) * 100

print("=== PERBANDINGAN FEATURE ENGINEERING ===")
print(f"Fitur baseline : {jumlah_fitur_baseline}")
print(f"Fitur refined  : {jumlah_fitur_refined}")
print(f"Pengurangan    : {pengurangan:.2f}%")
```

## Step 7 — Menetapkan Matriks TF-IDF Final

Hasil TF-IDF refined digunakan sebagai representasi numerik final dokumen karena memiliki jumlah fitur yang lebih terkendali setelah dilakukan penyaringan berdasarkan frekuensi term dan batas jumlah fitur.

Matriks ini akan digunakan pada tahap analisis berikutnya, seperti text similarity dan clustering.

```python
# Gunakan hasil refined sebagai matriks fitur final
X_final = X_tfidf_refined

# Nama fitur final
feature_names_final = feature_names_refined

print("=== MATRIKS FITUR FINAL ===")
print(f"Jumlah dokumen : {X_final.shape[0]}")
print(f"Jumlah fitur   : {X_final.shape[1]}")
print(f"Ukuran matriks : {X_final.shape}")
```

## Step 8 — Analisis Fitur Dominan pada Corpus

Analisis dilakukan terhadap total bobot TF-IDF setiap fitur pada seluruh dokumen.

Fitur dengan total bobot yang lebih tinggi menunjukkan term yang secara keseluruhan memiliki kontribusi lebih besar dalam representasi corpus.

```python
# Menjumlahkan bobot TF-IDF setiap fitur pada seluruh dokumen
total_tfidf = X_final.sum(axis=0).A1

# Membuat tabel fitur
top_terms = pd.DataFrame({
    'fitur': feature_names_final,
    'total_tfidf': total_tfidf
})

# Urutkan dari bobot terbesar
top_terms = top_terms.sort_values(
    by='total_tfidf',
    ascending=False
)

print("=== 20 FITUR DENGAN TOTAL BOBOT TF-IDF TERTINGGI ===")
display(top_terms.head(20).reset_index(drop=True))
```

## Step 9 — Integrasi Metadata dan Fitur TF-IDF

Matriks TF-IDF digabungkan dengan metadata dokumen agar setiap baris fitur tetap dapat ditelusuri ke artikel asalnya.

Kolom `topik` hanya berfungsi sebagai metadata untuk interpretasi dan evaluasi hasil analisis, bukan sebagai fitur dalam proses TF-IDF.

```python
# Ubah matriks TF-IDF menjadi DataFrame
df_tfidf_final = pd.DataFrame(
    X_final.toarray(),
    columns=feature_names_final
)

# Gabungkan metadata dengan fitur TF-IDF
df_matriks_final = pd.concat(
    [
        df[['no', 'judul', 'topik']].reset_index(drop=True),
        df_tfidf_final.reset_index(drop=True)
    ],
    axis=1
)

print("=== DATASET TF-IDF FINAL ===")
print(f"Jumlah baris : {df_matriks_final.shape[0]}")
print(f"Jumlah kolom : {df_matriks_final.shape[1]}")

display(df_matriks_final.iloc[:5, :10])
```

## Step 10 — Menyimpan Hasil Feature Engineering

Dataset hasil feature engineering disimpan dalam format CSV untuk digunakan pada tahap analisis berikutnya.

File berisi metadata dokumen dan fitur numerik TF-IDF refined.

```python
# Nama file output
nama_file_output = 'hasil_tfidf.csv'

# Simpan ke CSV
df_matriks_final.to_csv(
    nama_file_output,
    index=False
)

print(f"File berhasil disimpan sebagai: {nama_file_output}")
print(f"Ukuran dataset final: {df_matriks_final.shape}")
```

## Step 11 — Validasi Hasil Akhir

File hasil feature engineering dibaca kembali untuk memastikan proses penyimpanan berhasil dan struktur dataset tetap sesuai dengan matriks TF-IDF yang dihasilkan.

```python
# Baca kembali file hasil
df_check = pd.read_csv(nama_file_output)

print("=== VALIDASI FILE OUTPUT ===")
print(f"Jumlah baris : {df_check.shape[0]}")
print(f"Jumlah kolom : {df_check.shape[1]}")

print("\nKolom metadata:")
print(['no', 'judul', 'topik'])

print("\nJumlah fitur TF-IDF:")
print(len(df_check.columns) - 3)

print("\nCek nilai kosong:")
print(df_check[['no', 'judul', 'topik']].isnull().sum())

print("\n5 data pertama:")
display(df_check.head())
```
